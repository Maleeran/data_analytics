### pandas的数据结构介绍

#### Series

```python
import pandas as pd
obj = pd.Series([1, 2, 3, 4, 5])
obj
obj.index
```

```python
obj.values

obj.index #like range(5)
```

```python
obj2 = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
obj2
```

```python
obj2.index
```

```python
print(obj2['b'])
obj2['b'] = 7
obj2
```

```python
obj2[obj2 > 5]
```

```python
obj2[['a', 'c', 'd']]
```

```python
'g' in obj2
```

```python
'a' in obj2
```

可以直接通过字典来创建Series

```python
sdata = {'Ohio': 35000, 'Texas': 71000, 'Oregon': 16000, 'Utah': 5000}

obj3 = pd.Series(sdata)

obj3
```

```python
states = ['California', 'Ohio', 'Oregon', 'Texas']

obj4 =pd.Series(sdata, index=states)

obj4

pd.isnull(obj4)
```

```python
obj4['California'] = 45000
```
