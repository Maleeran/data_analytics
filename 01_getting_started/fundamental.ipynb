## 基础
### 核心数据类型
类型： int, float, string, bool, None

```python
int("100")  # 100
float("100")  # 100.0
```

### 容器结构

#### 列表 对应“一列数据”

```python
nums = [1,2,3,4,5]
nums[0]   # first
nums[-1]  # last
print(nums[-1])
nums[1:3] # slice from 1 to 3 (excluding 3)
nums[1:]  # slice from 1 to end
nums.append(6)
```
