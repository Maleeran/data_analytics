### 容器结构

#### 列表(list) 对应“一列数据”

```python
nums = [1,2,3,4,5]
nums[0]   # first
nums[-1]  # last
print(nums[-1])
nums[1:3] # slice from 1 to 3 (excluding 3)
nums[1:]  # slice from 1 to end
nums.append(6)
```

### 条件判断

```python
# if-else
age = 12
if age >= 18:
  print("your age is", age)
  print("Adult")
else:
  print("your age is", age)
  print("Teenager")
```

```python
# do-match
score = 'F' 

if score == 'A':
    print('Excellent')
elif score == 'B':
    print('Good')
elif score == 'C':
    print('Pass')
elif score == 'D':
    print('Fail')
else:
    print('Invalid score')
```

### 循环

```python
# for...in
for i in range(10):
    print(i)

names = ["Alice", "Bob", "Charlie"]    
for name in names:
    print(name)
```

```python
# do-while
sum = 0
while sum < 10:
    sum += 1
print(sum)
```

```python
L = ['Bart', 'Lisa', 'Adam']
for name in L:
  print('Hello, %s' % name)
```

```python
# break
n = 1
while n <= 100:
    if n > 10: # 当n = 11时，条件满足，执行break语句
        break # break语句会结束当前循环
    print(n)
    n = n + 1
print('END')
```

```python
# continue
n = 0
while n < 10:
    n = n + 1
    if n % 2 == 0: # 如果n是偶数，执行continue语句
        continue # continue语句会直接继续下一轮循环，后续的print()语句不会执行
    print(n)
```
