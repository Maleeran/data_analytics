###  字符串操作

- #### 字符串对象方法

```python
val = 'a, b, guido'
val.split(', ')
```

```python
pieces = [x.strip() for x in val.split(',')]

pieces
```

```python
first_item, second_item, third_item = pieces

first_item + '::' + second_item + '::' + third_item # 连接字符串
```

```python
'::'.join(pieces) # use `join` method to connect strings
```

```python
'guido' in val

val.index(',') # 返回第一个匹配的位置

val.find('guido')

val.find(';') # 找不到，返回-1
```

```python
val.count(',') # 计算字符串出现的次数
```
