#### 数据读取

- csv文件

```python
import pandas as pd
```

```python
file_path = 'data.csv'
try:
  df_csv = pd.read_csv(file_path,encoding = 'gbk')
  print("CSV 文件已读取成功")
  print(df_csv.head())
except FileNotFoundError:
  print("File not found")  
```

#### 数据查看与基本信息了解

```python
# 查看数据基本信息，包括列名、数据类型、非空值数量等
print(df_csv.info())

# 查看数据集行数，默认显示前 5 行
print(df_csv.head().to_csv(sep='\t', na_rep='nan'))

# 查看数据集行数，指定显示前 10 行
print(df_csv.head(10).to_csv(sep='\t', na_rep='nan'))

# 查看数据的基本统计信息，如均值、标准差、最小值、最大值等
print(df_csv.describe().to_csv(sep='\t', na_rep='nan'))

# 查看数据集行数
rows, columns = df_csv.shape

if rows < 1000:
    # 小数据集（行数少于1000）查看全量数据信息
    print(df_csv.to_csv(sep='\t', na_rep='nan'))
else:
    # 大数据集查看数据前几行信息
    print(df_csv.head().to_csv(sep='\t', na_rep='nan'))
```
