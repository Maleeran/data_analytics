- 滤除缺失数据

```python
from numpy import nan as NA

data = pd.Series([1, NA, 3.5, NA, 7])

data.dropna()
```

```python
data[data.notnull()] # 同上
```

```python
df = pd.DataFrame([[1., 6.5, 3.], [1., NA, NA], [NA, NA, NA], [NA, 6.5, 3.]])
```

```python
df
```

```python
df.dropna() # just one column's all rows aren't NA
```

```python
df.dropna(how='all')
```

```python
df[4] = NA
print(df)
df.dropna(axis=1, how='all') # 丢弃列，只需传入axis = 1
```

```python
df1 = pd.DataFrame(np.random.randn(7, 3))
df1.iloc[:4, 1] = NA
df1.iloc[:2, 2] = NA

df1
```

```python
df1.dropna()
```

```python
df1.dropna(thresh=2)
```

- #### 利用函数或映射进行数据转换

```python
data_map = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon', 'Pastrami', 'corned beef', 'Bacon', 'Bacon'], 'ounces': [4, 3, 12, 6, 7.5, 8, 9]})

data_map
```

```python
meat_to_animal = {'bacon': 'pig', 'pulled pork': 'pig', 'pastrami': 'cow', 'corned beef': 'cow', 'honey ham': 'pig', 'nova lox': 'salmon'}
```

```python
lowercased = data_map['food'].str.lower()

lowercased
```

```python
data_map['animal'] = lowercased.map(meat_to_animal)

data_map
```
