#### 数据清洗

- 处理缺失值

```python
import pandas as pd

csv_data = pd.read_csv("data.csv", encoding = 'gbk')
```

```python
# 删除包含缺失值的行
cleaned_data = csv_data.dropna()

# 用均值填充缺失值的行
for col in csv_data.columns:
    if csv_data[col].dtype in ['int64', 'float64']: # 找出数值型列
        csv_data[col].fillna(csv_data[col].mean(), inplace=True)

# 用中位数填充缺失值的行
for col in csv_data.columns:
    if csv_data[col].dtype in ['int64', 'float64']: # 找出数值型列
        csv_data[col].fillna(csv_data[col].median(), inplace=True)

# 用指定值填充缺失值
csv_data['column_name'].fillna('unknown', inplace=True)
```

- 处理重复值

```python
# 检查数据中是否有重复行
duplicate_rows = csv_data[csv_data.duplicated()]

# 删除重复行
csv_data = csv_data.drop_duplicates()
```

- 处理异常值

```python
import numpy as np

# 基于 Z-score 的异常值检测

z_scores = np.abs((csv_data - csv_data.mean())) / csv_data.std()
```
